```python
data=fetch_openml("car", as_frame=True).frame
data
```

```python
# ordinal encoder turn into numbers (that can fit machine algorithims)
columns_to_encode=["lug_boot", "safety"]

encoder=OrdinalEncoder(
    categories=[
        ["small", "med", "big"], 
        ["low", "med", "high"]
    ]
)

data[columns_to_encode]=encoder.fit_transform(data[columns_to_encode])
data
```

```python
# one hot encoder turns multiple values for an individual feature into multiple columns that can be 0 or 1
data=fetch_openml("adult", as_frame=True).frame
data
```

```python
data.occupation.value_counts()
```

```python
encoder=OneHotEncoder(handle_unknown="ignore", sparse_output=False)
encoded_value=encoder.fit_transform(data[["occupation", "race"]])
new_columns=encoder.get_feature_names_out(["occupation", "race"])
```

```python
encoded_value
```

```python
new_columns
```

```python
df_encoded=pd.DataFrame(encoded_value, columns=new_columns, index=data.index)
data_final=pd.concat(
    [data.drop(columns=["occupation", "race"]), df_encoded], axis=1
)
data_final
```
