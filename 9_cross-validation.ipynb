```python
from sklearn.datasets import load_breast_cancer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

```python
X, y=load_breast_cancer(return_X_y=True)
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.model_selection import cross_val_score
# split data into sections with hundreds of instances respectively
# validate with one section and train for the other sections
# repeat it and using different validation section
# dont need training and testing sets
clf=KNeighborsClassifier()
scores=cross_val_score(clf, X_scaled, y, cv=5, scoring='precision')
scores
```

```python
import numpy as np
np.mean(scores)
```
