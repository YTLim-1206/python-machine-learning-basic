```python
import seaborn as sns
```

```python
sns.set_theme(style="darkgrid")
```

```python
sns.color_palette()
```

```python
sns.set_palette(sns.color_palette("Paired"))
```

```python
sns.get_dataset_names()
```

```python
sns.load_dataset("tips")
```

```python
tips_data = sns.load_dataset("tips")
sns.scatterplot(tips_data, x="total_bill", y="tip", hue="sex", size="size", style="smoker")
```

```python
# more flexible (multiple graphs/axes, different kind......)
sns.relplot(tips_data, x="total_bill", y="tip", hue="sex", size="size", style="smoker", col="day", col_wrap=2, kind="scatter")
```

```python
# go to default 
sns.displot(tips_data, x="tip", col="sex")
```

```python
# replaces blocky bars of traditional histogram with smooth continuous probability density curve
sns.kdeplot(tips_data, x="tip", y="total_bill", hue="sex", fill=True)
# sns.displot(kind="kde"): recommended bcs more flexible
```

```python
sns.ecdfplot(tips_data, x="tip")
# sns.displot(kind="ecfd"): recommended bcs more flexible
```

```python
sns.displot(tips_data, x="tip", kind="hist" )
sns.rugplot(tips_data, x="tip", color="orange")
# shows actual individual value
```

```python
sns.catplot(tips_data, x="tip", y="day", kind="strip", hue="sex")
```

```python
sns.pairplot(tips_data)
# only use data with numbers and plot against each other
```

```python
sns.lmplot(tips_data, x="total_bill", y="tip")
```
