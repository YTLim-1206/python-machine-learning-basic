```python
# train classifier and make predictions
from sklearn.datasets import load_breast_cancer
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
X, y=load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test=train_test_split(X, y, test_size=0.2)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
# classifier method doesn't matter much
clf=KNeighborsClassifier()
# train the data
clf.fit(X_train_scaled, y_train)
# how well it performs
clf.score(X_test_scaled, y_test)
```

```python
single_instance=X_test_scaled[0]
single_instance
```

```python
# predict malignant/benign
clf.predict([single_instance])
```

```python
# see if prediction corrects
y_test[0]
```

```python
# predict probability: 40% 0, 60% 1
# not every classifier supports this
clf.predict_proba([single_instance])
```
